Write a Python program to get the number of people visiting a U.S. government website right now.


```python
import requests

# use url or link
url = 'https://analytics.usa.gov/data/live/realtime.json'

# request data
dispUsers = requests.get(url).json()

print(dispUsers['data'][0])
```

Write a Python program to test if a given page is found or not on the server.

```python
# import basic libraries to work with network
from urllib.request import Request, urlopen 
from urllib.error import HTTPError 
from urllib.error import URLError

url = 'https://www.firstpost.com/'
req = Request(url, headers={
  'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/118.0.0.0 Safari/537.36',
  'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8',
  'Accept-Language': 'en-US,en;q=0.5',
  'Referer': 'https://www.google.com/'
})

try:
  with urlopen(req, timeout=15) as resp:
    print(resp.status)
    body = resp.read(500)
    print(body)

except HTTPError as e:
  print(f'The error: {e}')

except URLError as e:
  print('Server not found')
```

Here, we use BeautifulSoup to parse the HTML content and format it in a clean, readable structure.
BeautifulSoup helps convert the raw HTML into a searchable tree of elements.

```python
import requests
from bs4 import BeautifulSoup

res = requests.get('https://www.w3resource.com/python-exercises/web-scraping/index.php')
soup = BeautifulSoup(res.content, 'html.parser')
print(soup.prettify())
```
