Write a Python program to create a lambda function that adds 15 to a given number passed in as an argument, also create a lambda function that multiplies argument x with argument y and prints the result.

```python
# pass only 1 argument and add 15 and return added value.
result = lambda a: a + 15
print(result(10))
```

```python
# pass 2 arguments and get their product
result2 = lambda x,y: x * y 
print(result2(12,4))
```

Write a Python program to create a function that takes one argument, and that argument will be multiplied with an unknown given number.

```python
def compute_function(n):
  return lambda x: x * n

result = compute_function(2)
print(f"Double the number of 15 = {result(15)}")
```

```python
result2 = compute_function(3)
print(f"Triple the number of 15 = {result2(15)}")
```

```python
result3 = compute_function(4)
print(f"Quadruple the number of 15 = {result3(15)}")
```

```python
result4 = compute_function(5)
print(f"Quintuple the number of 15 = {result4(15)}")
```

Write a Python program to sort a list of tuples using Lambda.

```python
# Create a list of tuples
subject_marks = [('English', 88), ('Science', 90), ('Math', 97), ('Social sciences', 82)]

print(subject_marks)

subject_marks.sort(key=lambda x: x[1])

print("\nSorting the list of tuples:")
print(subject_marks)
```
