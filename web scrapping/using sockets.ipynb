Most basic way to perform an HTTP request in Python: manually opening a TCP socket and then sending the HTTP request.

```python
import socket

HOST = 'www.udemy.com'
PORT = 80

client_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
client_socket.connect((HOST, PORT))

request = b"GET / HTTP/1.1\r\nHost: www.udemy.com\r\n\r\n"
client_socket.sendall(request)

response = ''
while True:
    recv = client_socket.recv(1024)
    if not recv:
        break
    response += recv.decode('utf-8', errors='ignore')

print(response)
client_socket.close()
```
